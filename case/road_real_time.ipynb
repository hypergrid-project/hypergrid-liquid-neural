# HyperGrid Real-Time Road Damage Detection (Case Study)
Notebook ini mendemonstrasikan proses inferensi data sensor jalan secara *real-time* menggunakan model **`hype_v2.onnx`** (dengan fitur Skor Confidence & Confidence Tiers) dan visualisasi peta spasial GIS.

```python
import os
import sys
sys.path.insert(0, os.path.abspath('..'))

import numpy as np
import pandas as pd
import folium
from folium.plugins import MarkerCluster, HeatMap
import onnxruntime as rt
from utils.pull_data import pull_road_data
from utils.features import RoadFeatureExtractorV2
from utils.trainer import get_confidence_tier

print('Dependencies and V2 modules loaded successfully.')
```

```python
# 1. Ambil data dari Supabase (atau fallback ke data sensor lokal jika offline)
data = pull_road_data()

if data and len(data) > 0:
    df_db = pd.DataFrame(data)
    print(f'Berhasil mengambil {len(df_db)} baris data dari Supabase.')
else:
    print('Menggunakan data sensor lokal sebagai simulasi data jalan...')
    fallback_file = '../data/Pothole/trip1_sensors.csv'
    if not os.path.exists(fallback_file):
        fallback_file = './data/Pothole/trip1_sensors.csv'
    df_db = pd.read_csv(fallback_file).head(500)
    print(f'Data simulasi lokal dimuat: {len(df_db)} baris.')

df_db.head()
```

```python
# 2. Preprocessing & Ekstraksi 47 Fitur Sensor Kausal V2 (Orientation-Invariant & Decomposed)
if 'sensor_x' in df_db.columns:
    df_db['accelerometerX'] = df_db['sensor_x']
    df_db['accelerometerY'] = df_db['sensor_y']
    df_db['accelerometerZ'] = df_db['sensor_z']
if 'speed_kmh' in df_db.columns:
    df_db['speed'] = df_db['speed_kmh']

extractor = RoadFeatureExtractorV2(window_size=15, step_size=3)
X_stream, metadata = extractor.extract_from_stream(df_db)

print(f'Matriks Fitur Input ONNX V2: {X_stream.shape}')
X_stream.head()
```

```python
# 3. Jalankan Inferensi Model ONNX V2
model_path = '../models/hype_v2.onnx'
if not os.path.exists(model_path):
    model_path = './models/hype_v2.onnx'

session = rt.InferenceSession(model_path)
input_name = session.get_inputs()[0].name
label_name = session.get_outputs()[0].name
prob_name = session.get_outputs()[1].name

inputs = X_stream.values.astype(np.float32)
preds, probs = session.run([label_name, prob_name], {input_name: inputs})

pothole_probs = probs[:, 1] if probs.ndim == 2 else probs
confidence_scores = np.where(preds == 1, pothole_probs, 1.0 - pothole_probs)
confidence_tiers = [get_confidence_tier(p) for p in pothole_probs]

metadata['predicted_label'] = preds
metadata['pothole_probability'] = pothole_probs
metadata['confidence'] = confidence_scores
metadata['confidence_tier'] = confidence_tiers
metadata['status'] = np.where(preds == 1, '⚠️ POTHOLE', '✅ NORMAL')

print('=== HASIL PREDIKSI REAL-TIME ONNX V2 ===')
print(metadata['status'].value_counts())
print('\nDistribusi Confidence Tier:')
print(metadata['confidence_tier'].value_counts())
metadata.head(10)
```

```python
# 4. Visualisasi Peta Spasial GIS dengan Tingkat Confidence (Folium)
df_map = metadata.dropna(subset=['latitude', 'longitude']).copy()
center_lat = df_map['latitude'].mean()
center_lon = df_map['longitude'].mean()

m = folium.Map(location=[center_lat, center_lon], zoom_start=15, tiles='OpenStreetMap')
cluster = MarkerCluster(name='Titik Deteksi').add_to(m)

for idx, row in df_map.iterrows():
    is_pothole = row['predicted_label'] == 1
    tier = row['confidence_tier']
    
    if is_pothole:
        color = '#c0392b' if tier == 'HIGH' else '#e67e22'
        title = f'⚠️ POTHOLE ({tier} CONFIDENCE)'
        radius = 7 if tier == 'HIGH' else 5
    else:
        color = '#3498db'
        title = '✅ NORMAL'
        radius = 3
    
    popup_content = f"""<b>{title}</b><hr>
    <b>Waktu:</b> {row['timestamp']}<br>
    <b>Pothole Prob:</b> {row['pothole_probability']:.2%}<br>
    <b>Confidence Score:</b> {row['confidence']:.2%}<br>
    <b>Peak Shock:</b> {row['peak_shock']:.2f} m/s²<br>
    <b>Kecepatan:</b> {row['speed']:.1f} km/h"""
    
    folium.CircleMarker(
        location=[row['latitude'], row['longitude']],
        radius=radius,
        color=color,
        fill=True,
        fill_color=color,
        fill_opacity=0.85 if is_pothole else 0.4,
        popup=folium.Popup(popup_content, max_width=260),
    ).add_to(cluster)

# Heatmap untuk Pothole
pothole_points = df_map[df_map['predicted_label'] == 1][['latitude', 'longitude', 'pothole_probability']].values.tolist()
if len(pothole_points) > 0:
    HeatMap(pothole_points, radius=15, blur=10).add_to(m)

folium.LayerControl().add_to(m)
m
```
